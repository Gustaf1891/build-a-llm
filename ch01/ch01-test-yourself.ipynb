### Questions by Chapter

#### 1.1 What is an LLM?

##### Q1: What is a large language model and how does it work?
A: An LLM is a deep learning network trained on massive amounts of text data to understand, generate and respond to human-like text. It follows a transformer structure which allows it to find relevant relationships and predicts words to generate text.

##### Q2: What is the significance of 'large' in 'large language models'?
A: It describes the large data size and model size.

##### Q3: How do LLMs relate to generative AI?
A: They are capable of generating text.

##### Q5: What is the difference between traditional machine learning and deep learning in terms of feature extraction?
A: They both extract features but for traditional machine learning the features have to be labeled by human experts, deep learning networks can learn features by themself. 

##### Q6: Match the terms to its description (page 5/6 in book)
A: LLM = A type of AI that uses deep learning to understand, generate and respond to human-like text. Transformer = An architecture used in LLMs that allows them to pay selective attention to different parts of the input when making predictions, making them adept at handling the nuances of human language. Generative AI = A type of artificial intelligence that can create new content, such as text, images, or audio. Deep Learning = A subset of machine learning that uses deep neural networks to model complex patterns and abstractions in data.

#### Stages of building and using LLMs

##### Q1: What are the main advantages of building custom LLMs compared to using general-purpose LLMs like ChatGPT?
A: They can be trained to specific tasks and perform better, they can also be used on personal data and be deployed on local devises reducing latency.

##### Q2: Describe the two-stage training process involved in creating an LLM.
A: Step one is to pre-train it on raw text data broad understanding of language, step two is to fine-tune it on smaller more specific data to adapt to particular tasks or domains.

##### Q3: What is the purpose of pretraining an LLM, and what type of data is used in this stage?
A: Pretraining allows the model to get a general understanding of language from a large & diverse dataset. It uses large amounts of raw text data to train the model to predict the next word in a sequence. 

##### Q4: Explain the concept of self-supervised learning in the context of pretraining LLMs.
A: Self-supervised learning is a method where the model generates labels for data on its own. This allows for training on large datasets without the need of human labels.

##### Q5: What are the two main categories of fine-tuning LLMs, and how do they differ in terms of the labeled data used?
A: Instruction fine-tuning uses instruction and answer pairs, such as translating text. Classification fine-tuning consits of text and class labels, e.g. emails labeled as spam or not spam.

##### Q6: Match the term to its description (page 8 in book)
A: Pretraining = The initial phase of training an LLM on a large, diverse dataset
to develop a broad understanding of language. Fine-tuning = The process of further training a pretrained LLM on a narrower dataset that is more specific to particular tasks or domains. Instruction fine-tuning = A type of fine-tuning where the labeled dataset consists of instruction and answer pairs, such as a query to translate a text
accompanied by the correctly translated text. Classification fine-tuning = A type of fine-tuning where the labeled dataset consists of texts and associated class labels, for example, emails associated with "spam" and "not spam" labels.
